# `render_html` local demo

Run all cells, then check the **inline iframe** below the last cell — it should play the episode with play/pause, a scrubber and a speed selector.

```python
import os
from env.env import DeliveryDrones
from env.wrappers import WindowedGridView
from agents.random import RandomAgent
from agents.dqn import DQNAgent, DenseQNetworkFactory
from helpers.rl_helpers import MultiAgentTrainer, render_html, ColabHTML

# Small, fast setup
env = WindowedGridView(DeliveryDrones(), radius=3)
env.env_params.update({'n_drones': 4})
env.reset()

agents = {d.index: RandomAgent(env) for d in env.drones}
agents[0] = DQNAgent(
    env, DenseQNetworkFactory(env, hidden_layers=[32, 32]),
    gamma=0.95, epsilon_start=1.0, epsilon_decay=0.999, epsilon_end=0.01,
    memory_size=10000, batch_size=64, target_update_interval=5,
)

# Quick train so drone 0 does something (bump this up for nicer behaviour)
MultiAgentTrainer(env, agents, reset_agents=True, seed=0).train(500)
print('trained')
```

```python
# Export the run as a self-contained HTML animation and show it inline
path = os.path.join('output', 'videos', 'local-demo.html')
render_html(env, agents, html_path=path, n_steps=100, fps=3, seed=None)
ColabHTML(path)
```
